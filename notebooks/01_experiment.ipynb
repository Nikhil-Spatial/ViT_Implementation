```python
Q_linear_transform = torch.nn.Linear(512, 128, bias=False)
K_linear_transform = torch.nn.Linear(512, 128, bias=False)
```

```python
inputs = torch.randn(256, 65, 512)
Q = Q_linear_transform(inputs)
K = K_linear_transform(inputs)
Q.shape, K.shape
```

```python
import math

output = (Q @ K.transpose(-2, -1)) / math.sqrt(128)
output.shape
```

```python
from torch.nn.functional import softmax

softmax(output, dim=-1)
```
